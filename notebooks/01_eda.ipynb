# EDA — Sleep-EDF SC processed epochs

Loads the real preprocessing manifest and plots class distribution.

```python
from pathlib import Path
import pandas as pd
from sleepstage.features.preprocessing import load_processed_npz
from sleepstage.visualization.training_curves import plot_class_distribution

manifest = pd.read_csv("../data/processed/manifest.csv")
manifest.head()
```

```python
counts = {s: 0 for s in ["W", "N1", "N2", "N3", "REM"]}
for _, row in manifest.iterrows():
    labels = load_processed_npz(Path(row["path"])).labels
    for i, s in enumerate(counts):
        counts[s] += int((labels == i).sum())
plot_class_distribution(counts, counts, Path("../reports/figures/03_class_distribution_nb.png"))
```
